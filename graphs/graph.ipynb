```python
import matplotlib.pyplot
import pandas
import numpy

import re
```

```python
fast_set_data : numpy.array = []
hash_set_data : numpy.array = []

with open("./../results/fast_set.txt", mode='r') as file:
    for line in file:
        if line.startswith("Solving MIS"):
            fast_set_data.append(float(re.search(r"[0-9]+.[0-9]+", line).group()))

with open("./../results/hash_set.txt", mode='r') as file:
    for line in file:
        if line.startswith("Solving MIS"):
            hash_set_data.append(float(re.search(r"[0-9]+.[0-9]+", line).group()))

print(fast_set_data)
print(hash_set_data)
```

```python
figure   = matplotlib.pyplot.figure()
axis     = figure.add_subplot()
box_plot = axis.bar(
    ["Fast Set", "Hash Set"],
    [numpy.median(fast_set_data), numpy.median(hash_set_data)])

axis.set_ylim(bottom=0.0)

matplotlib.pyplot.title("Median Processing Time")
matplotlib.pyplot.ylabel("Time (ms)")
```
